# 🎯 Intenções de Implementação (O Plano "Se-Então")

### 1. O Conceito: O que é (baseado em psicologia/ciência)
Desenvolvida pelo psicólogo Peter Gollwitzer, a teoria das **Intenções de Implementação** prova que metas vagas (como "vou focar mais" ou "vou tentar não me estressar") falham miseravelmente sob pressão. O cérebro precisa de instruções condicionais claras, como na programação.

A técnica consiste em pré-programar o seu comportamento usando a estrutura lógica **SE [gatilho/situação], ENTÃO [ação específica]**. 
Ao tomar a decisão de como agir *antes* da crise acontecer, você transfere o esforço da sua força de vontade (que é limitada) para o nível automático do cérebro. Você não precisa mais pensar no que fazer quando as coisas dão errado; você apenas executa o "script" pré-estabelecido.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
Depender da força de vontade no momento do caos (o *tilt*) consome uma energia mental absurda e quase sempre leva a decisões ruins.

* **Paralisia por Análise:** Ficar presa na Questão B de uma prova e perder 10 minutos preciosos apenas debatendo mentalmente: *"Será que eu continuo tentando essa ou pulo pra C? Mas eu já gastei tanto tempo aqui..."*.
* **Reações Emocionais:** Tomar um *Time Limit Exceeded* (TLE) e reagir com o fígado, mudando o código aleatoriamente e submetendo de novo em 30 segundos de puro desespero.
* **Procrastinação nos Estudos:** Sentar na frente do PC no fim de semana sem um plano claro. A falta de um "Se-Então" faz o cérebro escolher o caminho de menor resistência (abrir o YouTube ou o osu!) em vez de abrir o livro de algoritmos.

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Como Engenheira de Software, eu já sei escrever declarações `if-else` para as máquinas. Agora, escreverei para o meu próprio cérebro. Minhas regras de ouro para treinos e competições serão:

**Regra Acionável 1: O "Se-Então" do Tilt (Prevenção de Danos)**
* **SE** eu receber 2 *Wrong Answers* seguidos na mesma questão, **ENTÃO** eu tiro as mãos do teclado, levanto da cadeira e vou fazer carinho na minha gata (ou beber água) por exatos 2 minutos para ativar a quebra de padrão físico.

**Regra Acionável 2: O "Se-Então" do Tempo (Timeboxing)**
* **SE** o meu cronômetro bater 30 minutos em um problema e eu não tiver nenhuma ideia clara da lógica matemática por trás dele, **ENTÃO** eu paro imediatamente, leio o editorial ou recorro à teoria (como ler um capítulo do *Grokking Algorithms* sobre o assunto) para evitar a fadiga inútil.

**Regra Acionável 3: O "Se-Então" do Contest (Gestão de Prova)**
* **SE** eu ler uma questão na OBI/Codeforces e perceber que não consigo classificar a qual tag ela pertence em até 5 minutos, **ENTÃO** eu marco ela no rascunho, pulo para a próxima imediatamente e só volto nela no fim da prova.
