### 1. O Conceito: O que é (baseado em psicologia/ciência)
Na psicologia cognitiva, **Chunking** (ou Agrupamento) é o processo pelo qual o cérebro pega pequenas peças individuais de informação e as une em um único "bloco" (chunk) com significado. 

Nossa memória de trabalho (a "memória RAM" do cérebro) é muito limitada: conseguimos segurar apenas cerca de 4 a 7 itens isolados de uma vez. O *Chunking* é o "hack" do cérebro para burlar esse limite. 

**O Exemplo Clássico (Xadrez):** Quando um iniciante olha para um tabuleiro, ele vê 32 peças isoladas lutando por espaço — isso esgota o cérebro. Quando um Grande Mestre olha para o mesmo tabuleiro, ele não vê peças individuais; ele vê 3 ou 4 "blocos" de estratégias conhecidas (ex: "Defesa Siciliana com ataque no flanco"). O cérebro dele processa a mesma imagem gastando 10% da energia.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
Quando subimos de rating (como a transição do 800 para o 900 no Codeforces), os problemas deixam de ser "leia e implemente" e passam a exigir padrões. 

Se eu não tenho *chunks* formados na minha mente, cada problema de matemática, matriz ou array parece 100% novo e alienígena. 
Os sintomas de não usar o Chunking são:
* **Sobrecarga Cognitiva:** Ficar exausta após ler apenas uma ou duas questões, pois meu cérebro está tentando processar cada variável (`n`, `k`, `x`) de forma isolada.
* **Branco na hora de codar:** Olhar para a tela e não saber por onde começar a digitar.
* **Lentidão extrema:** Precisar reinventar a roda toda vez que preciso fazer uma Busca Binária ou achar o Maior Divisor Comum (MDC).

Isso suga a energia que eu precisaria usar para as disciplinas da faculdade, gerando estresse e a falsa sensação de que "não estou evoluindo".

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para parar de depender de decorar código e começar a construir *Chunks* visuais, vou aplicar as seguintes rotinas:

**Regra Acionável 1: Dar nome aos blocos**
Sempre que eu usar as dicas do TLE Eliminators ou ler um editorial, minha missão não é copiar o código, mas sim **dar um nome** ao truque que aprendi. 
* Em vez de pensar: *"Ah, eu tenho que somar de trás pra frente guardando o máximo..."*
* Eu crio o chunk: *"Isso é um problema de **Array de Sufixos**"*.
Dar um nome "empacota" a informação no cérebro.

**Regra Acionável 2: Ler código como blocos, não como linhas**
No meu template ou estudando, não leio mais `while (l <= r) { int mid = l + (r - l) / 2; ... }` caractere por caractere. Bato o olho nessa estrutura e meu cérebro deve ler apenas um grande bloco chamado `[BUSCA BINÁRIA]`.

**Rotina (Blind Re-solve):**
A única forma de forçar o cérebro a criar um *chunk* é a prática de recuperação. Se eu usei o editorial hoje para resolver um problema, eu salvo ele na pasta `revision`. Alguns dias depois, eu abro o problema e tento puxar o "bloco de lógica" da memória do zero, sem olhar a resposta anterior. Se eu conseguir, o *chunk* foi consolidado no meu "hardware".
