# 📅 Periodização de Treino (Macrociclos e Microciclos)

### 1. O Conceito: O que é (baseado em ciência do esporte)
A **Periodização Esportiva** é a técnica usada por atletas olímpicos para garantir que eles atinjam o pico absoluto de performance exatamente no dia da competição, sem sofrer *burnout* (esgotamento) no meio do caminho. 

Nenhum atleta treina na intensidade máxima o ano inteiro. O treinamento é dividido em ciclos:
* **Macrociclo:** O planejamento longo (ex: um semestre ou um ano inteiro).
* **Mesociclo:** Fases de 3 a 6 semanas com um objetivo específico (ex: "Focar apenas em construir base matemática e gulosos").
* **Microciclo:** O planejamento da semana atual.

O segredo da periodização é a fase de **Polimento (Tapering)**: semanas antes da competição principal, o volume de treino cai drasticamente para o corpo (e o cérebro) descansar e chegar 100% recuperado no dia da prova.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
Tentar tratar a Programação Competitiva como uma corrida de 100 metros rasos o ano inteiro é a receita perfeita para a frustração. 

* **Conflito de Calendários:** Tentar aprender um assunto denso e completamente novo de CP (como Grafos ou Programação Dinâmica) exatamente na mesma semana em que preciso entregar partes complexas de um trabalho da faculdade, como a definição de épicos e requisitos do Projeto Integrador. O cérebro simplesmente trava.
* **Mente Frita no Dia D:** Chegar na semana de competições importantes exausta porque passei os últimos 7 dias forçando a mente com problemas muito acima do meu rating. Na hora da prova, os padrões fáceis fogem da memória porque não houve o descanso do "Polimento".
* **Falta de "Off-Season":** Sentir culpa por ficar três dias sem abrir o Codeforces quando a faculdade exige atenção total.

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para garantir longevidade no esporte mental e excelência na Engenharia de Software, minha rotina não será fixa, mas sim adaptável ao longo dos meses:

**Regra Acionável 1: Sincronia de Macrociclos (Faculdade vs. CP)**
* **Fase de Base (Férias/Semanas Leves):** É aqui que eu gasto energia mental. Vou usar esse período para aprender algoritmos inéditos, errar bastante e expandir meu conhecimento teórico.
* **Fase de Manutenção (Semanas de Projetos/Provas):** Quando a carga da graduação estiver alta, a CP entra em modo de economia. Faço apenas uma ou duas questões confortáveis (rating 800-900) na semana inteira, só para não perder o ritmo de codificação, ou apenas leio editoriais no modo passivo.

**Regra Acionável 2: O Polimento para Competições**
Quando as datas de competições alvo (como a OBI ou a OPI Feminina) estiverem a 1 ou 2 semanas de distância, o treino muda completamente:
* **Proibido aprender assunto novo:** Tentar aprender um algoritmo do zero dias antes da prova só gera insegurança e consome a bateria mental.
* **Foco em Força:** Vou apenas revisar os *chunks* (padrões lógicos) que eu já domino, fazer problemas fáceis para ganhar confiança e velocidade de digitação, e dormir muito bem. O objetivo é chegar com a mente afiada, rápida e descansada.
