### 1. O Conceito: O que é (baseado em psicologia/ciência)
O termo "Post-Mortem" (pós-morte) é usado na medicina, na engenharia e nos e-Sports para descrever a análise técnica de um evento após a sua conclusão. 

Em esportes mentais de alto nível (como Xadrez e Poker), os jogadores têm uma regra de ouro: **nunca se analisa uma derrota no calor do momento**. Logo após perder uma partida, a amígdala cerebral está hiperativa (estado emocional alto) e o córtex pré-frontal (lógica) está suprimido. Analisar erros nesse estado gera autocrítica destrutiva. A Análise Post-Mortem é feita no dia seguinte, a frio, tratando o próprio erro não como uma falha pessoal, mas como um simples conjunto de dados a ser corrigido.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
Quando eu tomo um *Wrong Answer* no Teste 2 ou um *Time Limit Exceeded* (TLE) no Codeforces, a reação natural é o choque e a frustração.

* **Ego vs. Lógica:** Tentar debugar o código imediatamente após a falha geralmente vira uma batalha de ego. Eu fico tentando provar para o juiz online que a minha lógica inicial estava certa, em vez de procurar onde ela falhou.
* **A Falácia do Ajuste Cego:** Ficar mudando detalhes aleatórios no código (trocar `<` por `<=`, mudar o valor inicial de uma variável de $0$ para $1$) e submetendo várias vezes na esperança de "ficar verde". Isso destrói o rating e não ensina absolutamente nada.
* **Ciclo de Culpa:** Fechar o computador achando que "sou ruim nisso", quando na verdade o erro foi apenas um caso extremo (como um array só de zeros) que eu esqueci de tratar.

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para tirar a emoção da equação e evoluir como uma Engenheira de Software analítica, usarei a seguinte abordagem nas minhas sessões de treino e competições:

**Regra Acionável 1: A Quarentena do Código (Regra das 24 Horas)**
Se um problema me causou frustração ou *tilt*, eu não tenho permissão para olhar o editorial ou tentar reescrever o código no mesmo dia. 
* Vou colocar aquele problema em "quarentena".
* No dia seguinte, com o cérebro descansado e as emoções zeradas, eu abro o meu código que falhou e leio como se fosse o código de outra pessoa.

**Regra Acionável 2: O Interrogatório Objetivo**
Durante a análise fria, é proibido usar frases como "por que eu não pensei nisso?". O foco é puramente técnico. Vou fazer o seguinte checklist de 3 passos:
1. **Limites:** Houve *overflow*? Eu usei `int` onde a soma poderia passar de $2 \cdot 10^9$ e precisava de `long long`?
2. **Casos Extremos (Edge Cases):** Meu código funciona se $N = 1$? E se todos os números do vetor forem iguais?
3. **Complexidade:** Minha lógica tem dois loops aninhados (algo como $O(N^2)$) para um problema onde $N = 10^5$? Se sim, é óbvio que daria TLE, preciso achar uma forma gulosa de fazer em $O(N)$.

**Rotina de Resolução (O Diário de Bordo):**
Cada erro descoberto no Post-Mortem não é um fracasso, é uma vacina. Se eu descobrir que errei porque inicializei a variável de MDC com $1$ em vez de $0$, eu anoto isso mentalmente (ou no meu template). Da próxima vez, o erro não vai se repetir. O *Wrong Answer* fez o seu trabalho.
