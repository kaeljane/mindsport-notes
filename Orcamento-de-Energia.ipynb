### 1. O Conceito: O que é (baseado em psicologia/ciência)
A teoria do **Ego Depletion** (Esgotamento do Ego) e da **Fadiga de Decisão** sugere que a força de vontade e a capacidade analítica profunda não são infinitas. Elas funcionam exatamente como a bateria de um celular ou uma barra de "Mana" em um jogo de RPG.

Cada vez que tomamos uma decisão (mesmo as pequenas, como escolher o que vestir), resistimos a uma distração ou resolvemos um problema lógico complexo, gastamos um pouco dessa bateria. Quando a carga chega ao fim do dia no vermelho, o cérebro entra em modo de economia de energia: a qualidade das decisões despenca, a procrastinação domina e a força de vontade desaparece.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
O cérebro não separa a "energia de programar" da "energia de estudar". É o mesmo tanque de combustível. 

* **Conflito de Interesses:** A graduação em Engenharia de Software já exige uma carga cognitiva altíssima (arquitetura, projetos, matemática). Se eu gasto 80% da minha bateria diária lutando contra um problema *Greedy* no Codeforces, não vai sobrar energia para fazer um trabalho da faculdade à noite. O resultado não é preguiça, é falha de hardware (esgotamento).
* **Falso Desempenho:** Tentar estudar novos algoritmos ou fazer provas (como OBI) com a bateria no limite. Isso gera a sensação de que "eu não consigo aprender", quando na verdade o meu cérebro só não tem glicose e energia suficientes para registrar os *chunks* de informação.
* **Micro-decisões invisíveis:** Gastar energia mental preciosa decidindo "o que eu vou estudar hoje?", "qual questão vou fazer?" ou "como monto a estrutura base do código?". 

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para proteger a minha bateria mental e garantir que tanto a Engenharia de Software quanto a Programação Competitiva evoluam juntas, adotarei as seguintes estratégias:

**Regra Acionável 1: O "Cardápio" de Treino Baseado em Energia**
Não vou me prender a cronogramas rígidos que ignoram o meu cansaço real. Terei opções de treino ajustáveis:
* **Bateria Alta (Dias Livres):** Fazer *contests* simulados, resolver questões ligeiramente acima do meu rating (ex: 1000+) ou aprender um algoritmo totalmente novo.
* **Bateria Média:** Prática intercalada. Fazer questões de rating confortável (800-900) apenas para manter o reconhecimento de padrões afiado, sem fritar a mente.
* **Bateria Baixa (Semanas de Provas/Projetos densos):** Apenas leitura passiva. Abrir tutoriais, ler os *hints* do TLE Eliminators e estudar a lógica dos editoriais sem a pressão de codar nada.

**Regra Acionável 2: Redução de Atrito (Zero Micro-decisões)**
Vou automatizar tudo o que for possível para poupar energia para a lógica real do problema.
* Usar o meu template C++ padrão com todas as macros e bibliotecas já configuradas (o famoso `#include <bits/stdc++.h>` com os atalhos de `vector`).
* Deixar a lista de exercícios (a *CP-31 Sheet* ou o filtro do Codeforces) separada no dia anterior, para não gastar força de vontade escolhendo o que fazer na hora de sentar no computador.
