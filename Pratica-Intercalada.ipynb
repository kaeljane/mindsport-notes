### 1. O Conceito: O que é (baseado em ciência/psicologia)
A ciência da aprendizagem divide o estudo prático em duas abordagens principais:
* **Prática Blocada (Blocked Practice):** É como fomos ensinados na escola. Você aprende a fórmula de Bhaskara e depois faz 20 exercícios seguidos usando exatamente a mesma fórmula. Gera uma sensação rápida de fluência, mas péssima retenção a longo prazo.
* **Prática Intercalada (Interleaved Practice):** É misturar diferentes tipos de problemas em uma mesma sessão de estudos. O cérebro não sabe qual ferramenta vai usar a seguir. Isso é mais difícil e frustrante no curto prazo, mas ensina uma habilidade muito superior: a capacidade de **identificar** o problema, e não apenas de aplicar a solução.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
O maior erro na fase de base da Programação Competitiva é viciar em resolver listas filtradas por tópicos específicos.

* **A "Cegueira de Tag":** Se eu abro a tag *Greedy* no Codeforces e resolvo 5 problemas seguidos, a partir do segundo problema eu paro de ler a lógica. Eu apenas pego os arrays, ordeno e tento fazer uma simulação gulosa. Meu cérebro entra no piloto automático.
* **O Choque da Competição Real:** Nas provas da OBI, OPI Feminina ou em *contests* reais do Codeforces, as questões são "cegas" (não avisam o assunto). Se eu só treinei de forma blocada, quando vejo um problema sem tag, eu travo. Eu sei escrever o código de Busca Binária, mas não desenvolvi o "músculo" mental para perceber que aquele problema específico exigia Busca Binária.
* **Ilusão de Evolução:** Sentir que dominei Matemática Modular porque acertei 10 seguidas na mesma lista, mas errar feio quando um conceito modular aparece disfarçado em um problema de Grafos meses depois.

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para treinar o meu cérebro a reconhecer padrões (*Chunking*) de forma autônoma e me preparar para cenários reais de prova, mudarei a forma como escolho os problemas:

**Regra Acionável 1: O Treino "Cego" (Hide Tags)**
* Nas configurações do Codeforces, deixarei a opção de mostrar as *tags* dos problemas desligada por padrão.
* Mesmo se eu estiver seguindo uma planilha de estudos (como a *CP-31 Sheet*), farei o esforço consciente de ler o enunciado de forma neutra, formulando uma hipótese ("Acho que isso é matemática pura") antes de olhar a qual categoria a questão realmente pertence.

**Regra Acionável 2: A Salada de Frutas de Algoritmos**
Quando for montar minha sessão de treino (respeitando meu Orçamento de Energia), vou misturar os estímulos.
* Em vez de fazer 3 problemas de *Strings* no mesmo dia, farei 1 de *Strings*, 1 de *Math* e 1 de *Greedy*.
* Essa troca de contexto constante força a Metacognição a trabalhar e solidifica a minha capacidade de escolher a ferramenta certa do meu template C++ no meio da pressão.
