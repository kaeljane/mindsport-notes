# 🌱 Mentalidade de Crescimento vs. Fixa (Reframing)

![Topic](https://img.shields.io/badge/Topic-Psychology-purple?style=for-the-badge)
![Status](https://img.shields.io/badge/Status-Reviewed-success?style=for-the-badge)

### 1. O Conceito: O que é (baseado em psicologia/ciência)
A teoria da **Mentalidade (Mindset)**, desenvolvida pela psicóloga de Stanford Carol Dweck, divide a forma como enxergamos a nossa própria inteligência em duas categorias:

* **Mentalidade Fixa:** Acredita que a inteligência e o talento são traços inatos. Ou você nasce um "gênio da matemática", ou não nasce. O esforço é visto como um sinal de fraqueza (afinal, "se você precisa se esforçar tanto, é porque não tem o dom"). O fracasso é interpretado como uma prova de incompetência.
* **Mentalidade de Crescimento:** Acredita que o cérebro é como um músculo. A inteligência, o raciocínio lógico e a habilidade de programar podem ser desenvolvidos através de prática deliberada e boas estratégias. O fracasso não define quem você é; ele é apenas um *feedback* de que a estratégia atual não funcionou.

### 2. O Problema: Como a falta disso me afeta nos *contests* ou na faculdade
A cultura da tecnologia e da programação competitiva às vezes glorifica os "gênios", o que é um veneno para quem está construindo a sua base. Operar com uma Mentalidade Fixa gera comportamentos autossabotadores:

* **Fugir de Desafios:** Evitar tentar problemas com rating maior (ex: 1000+) pelo medo de não conseguir resolver e se sentir "burra". A Mentalidade Fixa prefere ficar na zona de conforto para proteger o ego.
* **O Peso do Veredito:** Levar um *Wrong Answer* (WA) no Codeforces para o lado pessoal, achando que isso significa que não serei uma boa Programadora Competitiva / Engenheira de Software no futuro.
* **Comparação Tóxica:** Olhar para competidores de alto nível (Grandmasters) e achar que eles operam em uma frequência inalcançável, ignorando as milhares de horas que eles passaram quebrando a cabeça e resolvendo o *Efeito Zeigarnik* de madrugada.

### 3. A Aplicação: Uma regra acionável ou rotina para resolver o problema

Para treinar o meu cérebro a amar o processo de aprendizado mais do que o resultado imediato, aplicarei a técnica do **Reframing** (Reestruturação Cognitiva):

**Regra Acionável 1: O Poder do "Ainda" (The Power of YET)**
O meu vocabulário interno dita os meus limites. Sempre que eu travar em um problema, a frase de autocrítica será imediatamente interceptada e reescrita:
* ❌ *Pensamento Fixo:* "Eu sou péssima em matemática modular, nunca vou subir para Pupil."
* ✅ *Pensamento de Crescimento:* "Eu não entendi esse padrão de matemática modular **ainda**."
A palavra *ainda* transforma um beco sem saída em uma linha de chegada futura.

**Regra Acionável 2: Ressignificando o *Wrong Answer***
Um erro no juiz online não é um atestado de burrice.
* A partir de agora, tratarei cada WA não como um fracasso, mas como **dados**. O juiz apenas me disse: *"Sua lógica não cobre o caso onde o array é negativo"*. É uma informação técnica, não uma ofensa pessoal. 

**Rotina de Foco no Esforço:**
Ao fim de um treino ou simulado, não vou me parabenizar apenas por ter resolvido a questão. O elogio interno será focado no processo: *"Gostei de como eu não desisti e usei a Análise Post-Mortem para achar a falha"*.
